```python
import nltk
import re
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from sklearn.feature_extraction.text import CountVectorizer
```

```python
nltk.download('punkt')
nltk.download('wordnet')
nltk.download('omw-1.4')  # open multilingual wordnet dataset
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
corpus = [
    "The cats were running faster than the dogs.",
    "Data science involves analyzing large datasets.",
    "Stemming and lemmatization are key NLP techniques."
]
```

```python
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))

def preprocess_text(text):
    # 1. Lowercase & Remove non-alphabetic characters
    text = re.sub('[^a-zA-Z]', ' ', text.lower())
    
    # 2. Tokenization: Split text into individual words
    tokens = word_tokenize(text)
    
    # 3. Stopword Removal, Stemming, and Lemmatization
    # Note: Usually you choose either Stemming OR Lemmatization, not both
    processed_tokens = [
        lemmatizer.lemmatize(word) # Converts word to base dictionary form
             for word in tokens 
             if word not in stop_words
          ]
    return " ".join(processed_tokens)

```

```python
cleaned_corpus = [preprocess_text(doc) for doc in corpus]
```

```python
vectorizer = CountVectorizer()
bow_matrix = vectorizer.fit_transform(cleaned_corpus)
```

```python
print("Cleaned Corpus:\n", cleaned_corpus)
print("\n")
print("Feature Names:\n", vectorizer.get_feature_names_out())
print("\n")
print("BoW Matrix:\n", bow_matrix.toarray())
```
